```python
#Q15. Ek complete "sanity check" function likho jo upar ke saare validation checks ek sath chalaye.
def sanitary_check(df):
    print('nulls values:',df.isnull().sum())
    print('duplicates values:',df.duplicated().sum())
    print('age range:',df['age'].min(),df['age'].max())
    print('credit score range:',df['credit_score'].min(),df['credit_score'].max())
    cals = (df['safe_driver_discount']+df['multi_policy_discount'] + df['bundling_discount']) * 50
    print('discount formula mismatch:',(cals != df['total_discounts']).sum())
sanitary_check(df)
```

```python
#Q50. Final Excel report banao: Conversion Summary by Source, Premium Summary by Region, Age Group Summary.
conv_summary = df.groupby('source_of_lead')['conversion_status'].mean().reset_index()
premium_summary = df.groupby('region')['premium_amount'].mean().reset_index()
age_summary = df.groupby('age_group')['premium_amount'].mean().reset_index()
with pd.ExcelWriter('insurance_excel.xlsx') as writer:
    df.to_excel(writer,sheet_name = 'clean data',index = False)
    conv_summary.to_excel(writer,sheet_name = 'conversion by source',index = False)
    premium_summary.to_excel(writer,sheet_name = 'premium by region ',index = False)
    age_summary.to_excel(writer,sheet_name = 'premium by age group', index = False)
print('saved')
```
